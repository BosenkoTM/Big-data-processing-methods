````python
#@title Ячейка 0. Мой вариант { display-mode: "form" }
MY_VARIANT = 25  #@param {type:"slider", min:1, max:25, step:1}

import pandas as pd
from IPython.display import display

PLATFORM_CYCLE = ['Arduino', 'Raspberry Pi', 'VEX', 'РОББО', 'Lego SPIKE']
AGE_CYCLE = ['Старшая (15-17)', 'Средняя (11-14)', 'Младшая (7-10)']
THRESHOLD_CYCLE = [60, 65, 70, 50, 55]
PARTITIONS_CYCLE = [16, 8, 2, 4]
SCALE_CYCLE = [1.0, 2.0, 3.0]
TASK_TITLES = ['Отсев и порог', 'Разрыв по этапам', 'Партиции и ядра', 'Возраст и ошибки', 'Точка перелома']

TASK_TEXTS = [
    "Для среза {selected_platform} × {selected_age_group} запишите долю отсева при пороге "
    "{final_score_threshold} и при пороге {thr_plus}. На сколько процентных пунктов вырос отсев? "
    "Какой порог вы предложили бы для школьного этапа и почему?",
    "Для среза {selected_platform} × {selected_age_group} при пороге {final_score_threshold} найдите "
    "этап с максимальным разрывом между «балл ≥ порога» и «прошли дальше» и объясните причину.",
    "Включите run_partition_sweep в ячейке 6. Сравните время при n_partitions = {n_partitions} "
    "с оптимальным числом партиций и объясните результат через число ядер Colab.",
    "Сравните долю участников с ≥ {high_error_level} ошибками на платформе {selected_platform} во всех "
    "трёх возрастных группах. С какого возраста вы рекомендуете эту платформу?",
    "В ячейке 5B при scale_max_millions = {scale_max_millions:g} найдите объём, при котором Spark "
    "обгоняет MapReduce, и число запросов, после которого окупается кэш. Сравните с моделью T = a + b·N.",
]
WHERE = {
    "VARIANT": "Ячейка 2",
    "final_score_threshold": "Ячейки 3 и 6",
    "N_SPLITS": "Ячейка 3",
    "n_partitions": "Ячейки 4 и 6",
    "scale_max_millions": "Ячейка 5B",
    "selected_platform": "Ячейка 6",
    "selected_age_group": "Ячейка 6",
    "high_error_level": "Ячейка 6",
}


def variant_plan(v):
    """Параметры варианта v (1…25); вариант 25 — эталон преподавателя."""
    return {
        "VARIANT": v,
        "final_score_threshold": THRESHOLD_CYCLE[v % 5],
        "N_SPLITS": 3 + v % 3,
        "n_partitions": PARTITIONS_CYCLE[v % 4],
        "scale_max_millions": SCALE_CYCLE[v % 3],
        "selected_platform": PLATFORM_CYCLE[(v // 5) % 5],
        "selected_age_group": AGE_CYCLE[v % 3],
        "high_error_level": 4 + v % 4,
    }


def variant_task(v):
    k = (v + v // 5) % 5
    p = variant_plan(v)
    return TASK_TITLES[k], TASK_TEXTS[k].format(thr_plus=p["final_score_threshold"] + 10, **p)


plan = variant_plan(MY_VARIANT)
display(pd.DataFrame({"Параметр": list(plan), "Значение": list(plan.values()),
                      "Где установить": [WHERE[k] for k in plan]}).style.hide(axis="index"))
title, text = variant_task(MY_VARIANT)
print(f"Вариант {MY_VARIANT}. Задание «{title}»:\n{text}")
````

````python
#@title Ячейка 1. Подготовка окружения { display-mode: "form" }
import os, re, sys, json, time, math, shutil, hashlib, platform, resource
import subprocess, importlib, statistics, tempfile, threading
from pathlib import Path


def pip_install(*pkgs):
    r = subprocess.run([sys.executable, "-m", "pip", "install", "-q", *pkgs],
                       capture_output=True, text=True)
    if r.returncode != 0:
        print(f"⚠ pip install {' '.join(pkgs)}: ошибка\n{r.stderr[-600:]}")
    return r.returncode == 0


# --- 1. PySpark ------------------------------------------------------------
try:
    import pyspark
except ImportError:
    pip_install("pyspark>=3.5,<4")
    importlib.invalidate_caches()
    import pyspark

# --- 2. Вспомогательные пакеты ----------------------------------------------
pip_install("findspark", "psutil")
pip_install("mrjob")          # может не установиться — это не критично
importlib.invalidate_caches()
import psutil

# --- 3. Java ------------------------------------------------------------------
def java_major_version():
    java_bin = "java"
    if os.environ.get("JAVA_HOME"):
        cand = Path(os.environ["JAVA_HOME"]) / "bin" / "java"
        if cand.exists():
            java_bin = str(cand)
    try:
        out = subprocess.run([java_bin, "-version"], capture_output=True, text=True)
    except FileNotFoundError:
        return None
    m = re.search(r'version "(\d+)(?:\.(\d+))?', out.stderr + out.stdout)
    if not m:
        return None
    major = int(m.group(1))
    return int(m.group(2)) if major == 1 else major


SPARK_MAJOR = int(pyspark.__version__.split(".")[0])
JAVA_OK = {17, 21} if SPARK_MAJOR >= 4 else {8, 11, 17}
JAVA_VERSION = java_major_version()
if JAVA_VERSION not in JAVA_OK:
    print(f"Java {JAVA_VERSION} не подходит для PySpark {pyspark.__version__} → устанавливаем OpenJDK 17…")
    subprocess.run("apt-get -qq update > /dev/null 2>&1 && "
                   "apt-get -qq install -y openjdk-17-jdk-headless > /dev/null 2>&1", shell=True)
    java_home = Path("/usr/lib/jvm/java-17-openjdk-amd64")
    if java_home.exists():
        os.environ["JAVA_HOME"] = str(java_home)
        os.environ["PATH"] = f"{java_home}/bin:" + os.environ["PATH"]
    JAVA_VERSION = java_major_version()
    if JAVA_VERSION not in JAVA_OK:
        print("⚠ Установите Java 17 вручную и перезапустите ячейку.")

# --- 4. findspark -------------------------------------------------------------
import findspark
try:
    findspark.init()
    FINDSPARK_INFO = findspark.find()
except Exception:
    FINDSPARK_INFO = "не требуется: PySpark установлен через pip"

# --- 5. SparkSession ----------------------------------------------------------
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import Window
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

spark = (SparkSession.builder
         .master("local[*]")                          # все ядра машины Colab
         .appName("BigData-Robotics-Lab")
         .config("spark.driver.memory", "4g")
         .config("spark.sql.shuffle.partitions", "8")  # вместо 200 по умолчанию
         .config("spark.ui.showConsoleProgress", "false")
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
sc = spark.sparkContext

# --- 6. mrjob -----------------------------------------------------------------
try:
    import mrjob
    MRJOB_OK, MRJOB_VERSION = True, mrjob.__version__
except Exception as e:
    MRJOB_OK, MRJOB_VERSION = False, f"недоступен ({type(e).__name__})"

# --- 7. Замер ресурсов --------------------------------------------------------
def nf(x):
    """Целое число с пробелом-разделителем разрядов: 50 000."""
    return f"{int(x):,}".replace(",", " ")


def to_mb(x):
    return None if x is None else x / 2**20


def dir_size(path):
    path = Path(path)
    if not path.exists():
        return 0
    if path.is_file():
        return path.stat().st_size
    return sum(p.stat().st_size for p in path.rglob("*") if p.is_file())


def _io_chars(p):
    io = p.io_counters()
    return getattr(io, "write_chars", io.write_bytes)


def run_monitored(cmd, shell=False, cwd=None, env=None, interval=0.02):
    """Запускает внешний процесс и измеряет время, пиковую RSS дерева процессов,
    объём записи (write-syscalls) и процессорное время дочерних процессов.
    Короткоживущие подпроцессы могут не попасть в выборку — оценки снизу."""
    out_f = tempfile.TemporaryFile(mode="w+", encoding="utf-8")
    err_f = tempfile.TemporaryFile(mode="w+", encoding="utf-8")
    ru0 = resource.getrusage(resource.RUSAGE_CHILDREN)
    t0 = time.perf_counter()
    proc = subprocess.Popen(cmd, shell=shell, cwd=cwd, env=env, stdout=out_f, stderr=err_f,
                            executable="/bin/bash" if shell else None)
    peak, io_last = 0, {}
    try:
        root = psutil.Process(proc.pid)
    except psutil.NoSuchProcess:
        root = None
    while proc.poll() is None:
        rss = 0
        try:
            procs = [root] + root.children(recursive=True)
        except Exception:
            procs = []
        for p in procs:
            try:
                rss += p.memory_info().rss
                io_last[p.pid] = _io_chars(p)
            except Exception:
                pass
        peak = max(peak, rss)
        time.sleep(interval)
    elapsed = time.perf_counter() - t0
    ru1 = resource.getrusage(resource.RUSAGE_CHILDREN)
    out_f.seek(0); err_f.seek(0)
    return {"returncode": proc.returncode, "stdout": out_f.read(), "stderr": err_f.read(),
            "elapsed": elapsed, "peak_rss": peak, "write_bytes": sum(io_last.values()),
            "cpu_sec": (ru1.ru_utime - ru0.ru_utime) + (ru1.ru_stime - ru0.ru_stime)}


class ResourceMonitor:
    """Фоновый замер пиковой RSS и объёма записи для текущего ядра Python и его
    дочерних процессов (в том числе JVM Spark)."""

    def __init__(self, interval=0.02):
        self.interval = interval
        self.peak_rss = 0
        self._io = {}
        self._stop = threading.Event()

    def _sample(self):
        me = psutil.Process(os.getpid())
        rss = 0
        for p in [me] + me.children(recursive=True):
            try:
                rss += p.memory_info().rss
                self._io[p.pid] = _io_chars(p)
            except Exception:
                pass
        self.peak_rss = max(self.peak_rss, rss)

    def _loop(self):
        while not self._stop.is_set():
            self._sample()
            time.sleep(self.interval)

    def __enter__(self):
        self._sample()
        self._io_start = dict(self._io)
        self._thread = threading.Thread(target=self._loop, daemon=True)
        self._thread.start()
        self.t0 = time.perf_counter()
        return self

    def __exit__(self, *exc):
        self.elapsed = time.perf_counter() - self.t0
        self._stop.set()
        self._thread.join()
        self._sample()
        self.write_bytes = sum(max(0, v - self._io_start.get(pid, 0)) for pid, v in self._io.items())
        return False


BENCH = globals().get("BENCH", {})   # сюда складываются результаты замеров

# --- 8. Сводка ------------------------------------------------------------------
print(f"Python        : {platform.python_version()}")
print(f"PySpark       : {pyspark.__version__}  (Spark UI: {sc.uiWebUrl})")
print(f"Java          : {JAVA_VERSION}")
print(f"findspark     : {FINDSPARK_INFO}")
print(f"mrjob         : {MRJOB_VERSION}")
print(f"CPU (логич.)  : {os.cpu_count()}   RAM: {psutil.virtual_memory().total / 2**30:.1f} ГБ")
print(f"Параллелизм Spark по умолчанию: {sc.defaultParallelism}")
````

## Ячейка 2. Генерация педагогического датасета

Датасет создаётся программно, поэтому работа не зависит от внешних ссылок. Одна строка — участие ученика в одном соревновании.

| Поле | Смысл |
|---|---|
| `student_id` | ID ученика; один ученик может участвовать в нескольких соревнованиях |
| `school_id` | ID школы или кружка |
| `age_group` | Младшая (7-10), Средняя (11-14), Старшая (15-17) |
| `robotics_platform` | Lego SPIKE, Arduino, РОББО, VEX, Raspberry Pi |
| `competition_stage` | Школьный, Муниципальный, Региональный, Всероссийский |
| `task_completion_time_sec` | время сборки и программирования, с |
| `telemetry_errors_count` | число ошибок калибровки датчиков и отладки кода |
| `final_score` | итоговый балл 0–100 |
| `qualified_next_stage` | 1 — прошёл на следующий этап (для Всероссийского — стал призёром) |

**Что заложено в генератор** (это знание пригодится при интерпретации):

* выбор платформы зависит от возраста: младшие чаще работают с Lego SPIKE и РОББО, старшие — с Arduino, VEX и Raspberry Pi;
* у платформ разная «трудность отладки»; для возраста 11–14 лет переход на Arduino и Raspberry Pi даёт дополнительный рост ошибок;
* у каждой школы свой «методический уровень», влияющий на ошибки и баллы;
* в нескольких школах заложен **аномальный профиль**: подозрительно быстрая сборка при большом числе ошибок. Список этих школ скрыт — его предстоит найти в задании со звёздочкой.

Файлы сохраняются в двух форматах: **CSV** (построчная обработка в MapReduce) и **JSON Lines** (демонстрация полуструктурированного формата). Затем CSV загружается в Spark с явной схемой.

````python
#@title Ячейка 2. Генерация датасета { display-mode: "form" }
VARIANT = 25  #@param {type:"slider", min:1, max:25, step:1}

import numpy as np
import pandas as pd
from IPython.display import display

SEED = 2026 + VARIANT
N_ROWS = 50_000 + ((VARIANT - 1) % 6) * 10_000      # 50 000 … 100 000 строк
N_SCHOOLS = 150
N_ANOMALY = 5 + VARIANT % 4                          # 5 … 8 «аномальных» школ
rng = np.random.default_rng(SEED)

PLATFORMS = ["Lego SPIKE", "Arduino", "РОББО", "VEX", "Raspberry Pi"]
AGE_GROUPS = ["Младшая (7-10)", "Средняя (11-14)", "Старшая (15-17)"]
STAGES = ["Школьный", "Муниципальный", "Региональный", "Всероссийский"]

# Вероятность выбора платформы в зависимости от возраста
PLATFORM_BY_AGE = {
    "Младшая (7-10)":  [0.45, 0.08, 0.35, 0.07, 0.05],
    "Средняя (11-14)": [0.25, 0.25, 0.20, 0.15, 0.15],
    "Старшая (15-17)": [0.10, 0.30, 0.10, 0.20, 0.30],
}
ERR_BASE = {"Lego SPIKE": 1.2, "Arduino": 3.0, "РОББО": 1.7, "VEX": 2.3, "Raspberry Pi": 3.3}
TIME_BASE = {"Lego SPIKE": 900, "Arduino": 1500, "РОББО": 1100, "VEX": 1300, "Raspberry Pi": 1700}
AGE_ERR = {"Младшая (7-10)": 1.25, "Средняя (11-14)": 1.0, "Старшая (15-17)": 0.75}
AGE_TIME = {"Младшая (7-10)": 1.20, "Средняя (11-14)": 1.0, "Старшая (15-17)": 0.90}
STAGE_TIME = {"Школьный": 1.0, "Муниципальный": 1.1, "Региональный": 1.2, "Всероссийский": 1.3}
STAGE_CUTOFF = {"Школьный": 60, "Муниципальный": 65, "Региональный": 72, "Всероссийский": 80}

# --- школы: размер и методический уровень ------------------------------------
school_ids = np.array([f"SCH-{k:03d}" for k in range(1, N_SCHOOLS + 1)])
school_weight = rng.gamma(2.0, 1.0, N_SCHOOLS)
school_weight /= school_weight.sum()
school_quality = rng.normal(0, 0.25, N_SCHOOLS)
big_schools = np.where(school_weight > np.median(school_weight))[0]
anomaly_idx = rng.choice(big_schools, N_ANOMALY, replace=False)
is_anomaly = np.zeros(N_SCHOOLS, dtype=bool)
is_anomaly[anomaly_idx] = True

# --- ученики: школа и возраст фиксированы ------------------------------------
n_students = N_ROWS // 2
st_school = rng.choice(N_SCHOOLS, n_students, p=school_weight)
st_age = rng.choice(len(AGE_GROUPS), n_students, p=[0.35, 0.40, 0.25])

# --- участия в соревнованиях --------------------------------------------------
row_student = rng.integers(0, n_students, N_ROWS)
df = pd.DataFrame({
    "student_id": np.char.add("ST-", np.char.zfill(row_student.astype(str), 6)),
    "school_idx": st_school[row_student],
    "age_group": np.array(AGE_GROUPS)[st_age[row_student]],
})
df["school_id"] = school_ids[df["school_idx"].to_numpy()]
df["robotics_platform"] = ""
for age, probs in PLATFORM_BY_AGE.items():
    mask = df["age_group"] == age
    df.loc[mask, "robotics_platform"] = rng.choice(PLATFORMS, mask.sum(), p=probs)
df["competition_stage"] = rng.choice(STAGES, N_ROWS, p=[0.50, 0.28, 0.16, 0.06])

q = school_quality[df["school_idx"].to_numpy()]
anom = is_anomaly[df["school_idx"].to_numpy()]
middle_hard = ((df["age_group"] == "Средняя (11-14)")
               & df["robotics_platform"].isin(["Arduino", "Raspberry Pi"])).to_numpy()

lam = (df["robotics_platform"].map(ERR_BASE).to_numpy()
       * df["age_group"].map(AGE_ERR).to_numpy()
       * np.exp(-q)
       * np.where(middle_hard, 1.25, 1.0)
       * np.where(anom, 2.3, 1.0))
errors = rng.poisson(lam)

expected_time = (df["robotics_platform"].map(TIME_BASE).to_numpy()
                 * df["age_group"].map(AGE_TIME).to_numpy()
                 * df["competition_stage"].map(STAGE_TIME).to_numpy())
time_sec = expected_time * np.exp(rng.normal(0, 0.35, N_ROWS)) * np.where(anom, 0.5, 1.0)
time_sec = np.clip(np.round(time_sec), 120, 7200).astype(int)

score = (80 + 10 * q - 3.2 * errors
         - 8 * np.log(time_sec / expected_time)
         + rng.normal(0, 8, N_ROWS))
score = np.clip(np.round(score), 0, 100).astype(int)

cutoff = df["competition_stage"].map(STAGE_CUTOFF).to_numpy()
qualified = (score + rng.normal(0, 3, N_ROWS) >= cutoff).astype(int)

df["task_completion_time_sec"] = time_sec
df["telemetry_errors_count"] = errors
df["final_score"] = score
df["qualified_next_stage"] = qualified

COLUMNS = ["student_id", "school_id", "age_group", "robotics_platform", "competition_stage",
           "task_completion_time_sec", "telemetry_errors_count", "final_score",
           "qualified_next_stage"]
df = df[COLUMNS].reset_index(drop=True)

# --- сохранение ----------------------------------------------------------------
DATA_DIR = Path("data"); DATA_DIR.mkdir(exist_ok=True)
CSV_PATH = DATA_DIR / "robotics_competitions.csv"
JSON_PATH = DATA_DIR / "robotics_competitions.jsonl"
KEY_PATH = DATA_DIR / "_answer_key.json"

df.to_csv(CSV_PATH, index=False, encoding="utf-8")
df.to_json(JSON_PATH, orient="records", lines=True, force_ascii=False)

# Ключ для автотеста: хранятся только хэши ID аномальных школ
salt = hashlib.sha256(f"robotics-{SEED}".encode()).hexdigest()[:16]
KEY_PATH.write_text(json.dumps({
    "salt": salt,
    "n_anomalies": int(N_ANOMALY),
    "anomaly_hashes": sorted(hashlib.sha256((salt + s).encode()).hexdigest()
                             for s in school_ids[anomaly_idx]),
}), encoding="utf-8")

print(f"Вариант {VARIANT}: {nf(len(df))} строк, {nf(df['student_id'].nunique())} учеников, "
      f"{df['school_id'].nunique()} школ; аномальных школ заложено: {N_ANOMALY}")
print(f"CSV  : {CSV_PATH}  ({to_mb(CSV_PATH.stat().st_size):.1f} МБ)")
print(f"JSONL: {JSON_PATH} ({to_mb(JSON_PATH.stat().st_size):.1f} МБ)")
display(df.head(8))
display(df.describe().round(2))

# --- загрузка в Spark с явной схемой ------------------------------------------------
SCHEMA = StructType([
    StructField("student_id", StringType(), False),
    StructField("school_id", StringType(), False),
    StructField("age_group", StringType(), False),
    StructField("robotics_platform", StringType(), False),
    StructField("competition_stage", StringType(), False),
    StructField("task_completion_time_sec", IntegerType(), False),
    StructField("telemetry_errors_count", IntegerType(), False),
    StructField("final_score", IntegerType(), False),
    StructField("qualified_next_stage", IntegerType(), False),
])


def read_csv_df():
    """Каждый вызов создаёт новый план чтения CSV с диска (без кэша)."""
    return spark.read.csv(str(CSV_PATH), header=True, schema=SCHEMA, encoding="UTF-8")


sdf = read_csv_df()
sdf_json = spark.read.schema(SCHEMA).json(str(JSON_PATH))
print(f"Spark: CSV → {nf(sdf.count())} строк, JSONL → {nf(sdf_json.count())} строк; "
      f"партиций при чтении CSV: {sdf.rdd.getNumPartitions()}")
sdf.printSchema()
````

## Ячейка 3. Модуль Hadoop MapReduce

### Теория: почему MapReduce «живёт» на диске

Hadoop создавался для кластеров из тысяч недорогих машин, где отказ узла — обычное событие. Отсюда архитектурные решения:

1. **HDFS** хранит файлы блоками (по умолчанию 128 МБ), каждый блок реплицируется (обычно ×3). Вычисление отправляется к данным (*data locality*), а не данные к вычислению.
2. **Map-фаза.** Каждый блок (input split) обрабатывает отдельный mapper, который выдаёт пары `ключ → значение`. Выход mapper **сбрасывается на локальный диск** узла (spill), сортируется и разбивается по reducer.
3. **Combiner** (необязательно) — «мини-reducer» на стороне mapper: агрегирует пары до передачи по сети и уменьшает объём shuffle.
4. **Shuffle & Sort.** Пары с одинаковым ключом собираются на одном reducer, данные передаются по сети и снова пишутся на диск.
5. **Reduce-фаза.** Reducer агрегирует значения и **записывает результат в HDFS**.

Промежуточная материализация на диск делает систему устойчивой: упавшую задачу можно перезапустить с сохранённого входа. Цена — много операций ввода-вывода, особенно в итеративных алгоритмах (машинное обучение, графы), где каждый шаг снова читает и пишет диск.

```
 CSV (HDFS) ─► split 1 ─► mapper ─► sort ─► combiner ─┐
            ─► split 2 ─► mapper ─► sort ─► combiner ─┼─► shuffle/sort ─► reducer ─► part-00000 (HDFS)
            ─► split N ─► mapper ─► sort ─► combiner ─┘      (диск)                     (диск)
```

### Задача

Для каждой робототехнической платформы посчитать **средний балл**, **среднее число ошибок отладки**, **долю участников, набравших не меньше порога**, и **число участий**.

**Важная деталь.** Combiner передаёт дальше **суммы и счётчики**, а не средние. Среднее от средних по разным разбиениям даёт неверный результат, если разбиения разного размера.

### Два способа запуска

* **`mrjob`, runner `local`** — класс `MRJob` с методами `mapper`, `combiner`, `reducer`. Runner разбивает вход, запускает задачи отдельными процессами и хранит промежуточные файлы во временном каталоге. Тот же код без изменений запускается на кластере (`-r hadoop`) и в облаке (`-r emr`).
* **Hadoop Streaming (эмуляция)** — те же стадии на отдельных Python-скриптах, связанных через stdin/stdout и `sort`. Именно так работает `hadoop-streaming.jar`. Этот путь выполняется всегда и служит резервным, если `mrjob` не установился.

````python
#@title Ячейка 3. MapReduce: mrjob + Hadoop Streaming { display-mode: "form" }
final_score_threshold = 60  #@param {type:"slider", min:0, max:100, step:5}
N_SPLITS = 4                #@param {type:"slider", min:1, max:8, step:1}

THR = float(final_score_threshold)
env = dict(os.environ, PYTHONIOENCODING="utf-8", LC_ALL="C")

# Python 3.13 удалил модуль pipes, который импортирует mrjob 0.7.4.
# Подключаем совместимый модуль через PYTHONPATH; код задания не меняется.
if sys.version_info >= (3, 13):
    SHIM_DIR = Path("compat_shims").resolve()
    SHIM_DIR.mkdir(exist_ok=True)
    (SHIM_DIR / "pipes.py").write_text("from shlex import quote\n__all__ = ['quote']\n",
                                       encoding="utf-8")
    env["PYTHONPATH"] = os.pathsep.join(p for p in (str(SHIM_DIR), env.get("PYTHONPATH")) if p)

# ============ 3.1. Задание MRJob ==============================================
MRJOB_CODE = r"""
from mrjob.job import MRJob


class MRPlatformStats(MRJob):
    # Средний балл, среднее число ошибок и доля прошедших порог по платформе

    def configure_args(self):
        super().configure_args()
        self.add_passthru_arg("--threshold", type=float, default=60.0)

    def mapper(self, _, line):
        if line.startswith("student_id"):          # пропускаем заголовок CSV
            return
        f = line.split(",")
        score = float(f[7])
        errors = int(f[6])
        passed = 1 if score >= self.options.threshold else 0
        yield f[3], (score, errors, passed, 1)      # ключ — платформа

    def combiner(self, platform, values):
        s = e = p = n = 0
        for score, errors, passed, cnt in values:
            s += score; e += errors; p += passed; n += cnt
        yield platform, (s, e, p, n)                # суммы, а не средние!

    def reducer(self, platform, values):
        s = e = p = n = 0
        for score, errors, passed, cnt in values:
            s += score; e += errors; p += passed; n += cnt
        yield platform, {"avg_score": round(s / n, 4), "avg_errors": round(e / n, 4),
                         "pass_rate": round(p / n, 4), "n": n}


if __name__ == "__main__":
    MRPlatformStats.run()
"""
Path("mr_platform_stats.py").write_text(MRJOB_CODE, encoding="utf-8")


def parse_mr_output(text):
    rows = []
    for line in text.strip().splitlines():
        if "\t" not in line:
            continue
        k, v = line.split("\t", 1)
        try:
            k = json.loads(k)          # mrjob выводит ключ в JSON-кавычках
        except json.JSONDecodeError:
            pass                       # Streaming — ключ как есть
        rows.append({"robotics_platform": k, **json.loads(v)})
    return pd.DataFrame(rows).sort_values("robotics_platform").reset_index(drop=True)


mr_result = None
if MRJOB_OK:
    MR_TMP = Path("mr_tmp")
    shutil.rmtree(MR_TMP, ignore_errors=True)
    attempts = [
        ("local", ["-r", "local", "--local-tmp-dir", str(MR_TMP), "--cleanup", "NONE"]),
        ("local", ["-r", "local"]),
        ("inline", ["-r", "inline"]),
    ]
    for runner, opts in attempts:
        cmd = [sys.executable, "mr_platform_stats.py", *opts,
               "--threshold", str(THR), str(CSV_PATH)]
        run = run_monitored(cmd, env=env)
        if run["returncode"] == 0 and "\t" in run["stdout"]:
            mr_result = parse_mr_output(run["stdout"])
            tmp_bytes = dir_size(MR_TMP)
            BENCH["MapReduce: mrjob"] = {
                "family": "MapReduce", "time_s": run["elapsed"], "runs": [run["elapsed"]],
                "peak_rss_mb": to_mb(run["peak_rss"]), "write_mb": to_mb(run["write_bytes"]),
                "tmp_files_mb": to_mb(tmp_bytes) if tmp_bytes else None,
                "cpu_s": run["cpu_sec"], "note": f"runner={runner}"}
            print(f"mrjob (runner={runner}): {run['elapsed']:.2f} с, "
                  f"промежуточные файлы: {to_mb(tmp_bytes):.2f} МБ")
            break
        err_lines = [l for l in run["stderr"].splitlines() if l.strip()]
        print(f"⚠ mrjob runner={runner}: код {run['returncode']}: "
              f"{err_lines[-1] if err_lines else 'нет сообщения'}")
    shutil.rmtree(MR_TMP, ignore_errors=True)
else:
    print("mrjob недоступен — используем только Hadoop Streaming.")

# ============ 3.2. Hadoop Streaming (эмуляция) ====================================
Path("mapper.py").write_text(r"""
import sys
thr = float(sys.argv[1])
for line in sys.stdin:
    f = line.rstrip("\n").split(",")
    if len(f) != 9 or f[0] == "student_id":
        continue
    score = float(f[7])
    print(f"{f[3]}\t{score}\t{int(f[6])}\t{int(score >= thr)}\t1")
""", encoding="utf-8")

AGG_CODE = r"""
import sys, json
MODE = sys.argv[1]            # combine | reduce
def emit(key, a):
    if key is None:
        return
    if MODE == "combine":
        print(key, a[0], a[1], a[2], a[3], sep="\t")
    else:
        n = a[3]
        print(key, json.dumps({"avg_score": round(a[0] / n, 4), "avg_errors": round(a[1] / n, 4),
                               "pass_rate": round(a[2] / n, 4), "n": n}), sep="\t")
cur, acc = None, [0.0, 0, 0, 0]
for line in sys.stdin:
    key, s, e, p, n = line.rstrip("\n").split("\t")
    if key != cur:                 # вход отсортирован: новый ключ — новая группа
        emit(cur, acc)
        cur, acc = key, [0.0, 0, 0, 0]
    acc[0] += float(s); acc[1] += int(e); acc[2] += int(p); acc[3] += int(n)
emit(cur, acc)
"""
Path("aggregate.py").write_text(AGG_CODE, encoding="utf-8")

Path("streaming_job.sh").write_text(r"""#!/usr/bin/env bash
set -euo pipefail
INPUT="$1"; THR="$2"; SPLITS="$3"; WORK="$4"; PY="$5"
rm -rf "$WORK"; mkdir -p "$WORK/splits" "$WORK/spill" "$WORK/map" "$WORK/out"
# 1) «загрузка в HDFS» и нарезка на input splits
tail -n +2 "$INPUT" > "$WORK/input.csv"
split -n l/"$SPLITS" -d "$WORK/input.csv" "$WORK/splits/part-"
# 2) map (выход сбрасывается на локальный диск) → сортировка → combiner;
#    все splits обрабатываются параллельно
pids=()
for f in "$WORK"/splits/part-*; do
  name=$(basename "$f")
  ( "$PY" mapper.py "$THR" < "$f" > "$WORK/spill/$name" \
    && sort "$WORK/spill/$name" | "$PY" aggregate.py combine > "$WORK/map/$name" ) &
  pids+=($!)
done
for p in "${pids[@]}"; do wait "$p"; done
# 3) shuffle & sort: объединяем выходы всех mapper и сортируем по ключу
sort "$WORK"/map/part-* > "$WORK/shuffle.txt"
# 4) reduce → результат на «HDFS»
"$PY" aggregate.py reduce < "$WORK/shuffle.txt" > "$WORK/out/part-00000"
""", encoding="utf-8")

WORK = Path("streaming_work")
cmd = f'bash streaming_job.sh "{CSV_PATH}" {THR} {N_SPLITS} "{WORK}" "{sys.executable}"'
run = run_monitored(cmd, shell=True, env=env)
if run["returncode"] != 0:
    raise RuntimeError(run["stderr"])
streaming_result = parse_mr_output((WORK / "out" / "part-00000").read_text(encoding="utf-8"))
stage_sizes = {
    "input + splits (HDFS)": dir_size(WORK / "input.csv") + dir_size(WORK / "splits"),
    "выход mapper (spill на диск)": dir_size(WORK / "spill"),
    "после combiner": dir_size(WORK / "map"),
    "shuffle/sort": dir_size(WORK / "shuffle.txt"),
    "output (HDFS)": dir_size(WORK / "out"),
}
BENCH["MapReduce: Streaming"] = {
    "family": "MapReduce", "time_s": run["elapsed"], "runs": [run["elapsed"]],
    "peak_rss_mb": to_mb(run["peak_rss"]), "write_mb": to_mb(run["write_bytes"]),
    "tmp_files_mb": to_mb(sum(stage_sizes.values())), "cpu_s": run["cpu_sec"],
    "note": f"{N_SPLITS} splits"}
print(f"Hadoop Streaming: {run['elapsed']:.2f} с; файлы по стадиям:")
for k, v in stage_sizes.items():
    print(f"   {k:<32} {v / 1024:>10.1f} КБ")

# ============ 3.3. Эталон на pandas и сверка ==========================================
t0 = time.perf_counter()
pandas_ref = (df.assign(passed=(df["final_score"] >= THR).astype(int))
                .groupby("robotics_platform")
                .agg(avg_score=("final_score", "mean"),
                     avg_errors=("telemetry_errors_count", "mean"),
                     pass_rate=("passed", "mean"),
                     n=("final_score", "size"))
                .round(4).reset_index())
t_pandas = time.perf_counter() - t0
BENCH["Pandas (эталон)"] = {"family": "Pandas", "time_s": t_pandas, "runs": [t_pandas],
                            "peak_rss_mb": None, "write_mb": 0.0, "tmp_files_mb": 0.0,
                            "cpu_s": None, "note": "1 процесс, данные уже в памяти"}


def max_diff(res, ref=pandas_ref):
    m = res.merge(ref, on="robotics_platform", suffixes=("", "_ref"))
    assert len(m) == len(ref), "набор платформ не совпадает с эталоном"
    assert (m["n"] == m["n_ref"]).all(), "число строк по платформам не совпадает"
    return max((m[c] - m[c + "_ref"]).abs().max() for c in ["avg_score", "avg_errors", "pass_rate"])


for name, res in [("mrjob", mr_result), ("Streaming", streaming_result)]:
    if res is not None:
        print(f"Сверка {name:<9} с pandas: максимальное расхождение = {max_diff(res):.2e}")

print(f"\nРезультат MapReduce (порог = {THR:g}):")
display(streaming_result if mr_result is None else mr_result)
````

## Ячейка 4. Модуль Apache Spark

### Теория

* **RDD** (Resilient Distributed Dataset) — неизменяемая распределённая коллекция объектов, разбитая на **партиции**. Устойчивость обеспечивается не копиями данных, а **lineage**: Spark помнит цепочку преобразований и при сбое пересчитывает только потерянные партиции.
* **DataFrame** — RDD со схемой (именованные типизированные столбцы). Запрос проходит через оптимизатор **Catalyst**: логический план → оптимизированный план (перенос фильтров к источнику, отбор нужных столбцов) → физический план. Код исполняется движком **Tungsten** в компактном бинарном формате.
* **Spark SQL** — тот же DataFrame, записанный на SQL; план выполнения совпадает.
* **Ленивые вычисления (Lazy Evaluation).** Преобразования (`filter`, `groupBy`, `map`) только достраивают план. Вычисление запускает **действие** (`collect`, `count`, `show`, `write`).
* **DAG и In-Memory.** Spark строит направленный ациклический граф стадий. Узкие преобразования объединяются в одну стадию без промежуточной записи. Граница стадии — shuffle (например, `groupBy`): shuffle-файлы Spark тоже пишет на диск, но между операциями данные остаются в памяти, а `cache()` позволяет повторно использовать уже прочитанный набор без обращения к диску.
* **Партиции** — единица параллелизма: одна партиция обрабатывается одной задачей на одном ядре. Слишком мало партиций — простаивают ядра; слишком много — накладные расходы на планирование превышают выигрыш.

### Что измеряем

| Режим | Смысл |
|---|---|
| **холодный** | каждый запрос заново читает CSV с диска — аналог MapReduce |
| **в памяти (`cache`)** | данные один раз прочитаны и закэшированы; так работает интерактивная аналитика |

Каждый замер повторяется `REPEATS` раз, в таблицу идёт **медиана**. Первый запуск после старта JVM обычно заметно медленнее из-за «прогрева».

````python
#@title Ячейка 4. Spark: RDD, DataFrame, SQL { display-mode: "form" }
n_partitions = 8  #@param {type:"slider", min:1, max:64, step:1}
REPEATS = 3       #@param {type:"slider", min:1, max:10, step:1}

THR = float(final_score_threshold)
spark.conf.set("spark.sql.shuffle.partitions", str(n_partitions))
# чтобы при «холодном» чтении CSV получилось ~n_partitions разбиений
csv_bytes = CSV_PATH.stat().st_size
spark.conf.set("spark.sql.files.maxPartitionBytes", str(max(64 * 1024, math.ceil(csv_bytes / n_partitions))))


# ============ 4.1. Три способа решения одной задачи ============================
def spark_df_stats(frame, threshold):
    """DataFrame API"""
    return (frame.groupBy("robotics_platform")
                 .agg(F.round(F.avg("final_score"), 4).alias("avg_score"),
                      F.round(F.avg("telemetry_errors_count"), 4).alias("avg_errors"),
                      F.round(F.avg((F.col("final_score") >= threshold).cast("int")), 4).alias("pass_rate"),
                      F.count(F.lit(1)).alias("n"))
                 .orderBy("robotics_platform"))


def spark_sql_stats(frame, threshold):
    """Spark SQL"""
    frame.createOrReplaceTempView("robotics")
    return spark.sql(f"""
        SELECT robotics_platform,
               ROUND(AVG(final_score), 4)                                   AS avg_score,
               ROUND(AVG(telemetry_errors_count), 4)                        AS avg_errors,
               ROUND(AVG(CASE WHEN final_score >= {float(threshold)} THEN 1 ELSE 0 END), 4) AS pass_rate,
               COUNT(*)                                                     AS n
        FROM robotics
        GROUP BY robotics_platform
        ORDER BY robotics_platform
    """)


def parse_line(line):
    f = line.split(",")
    return f[3], float(f[7]), int(f[6])


def spark_rdd_stats(rdd_lines, threshold):
    """RDD API: те же map и reduce, что в MapReduce"""
    pairs = (rdd_lines
             .map(lambda t, thr=threshold: (t[0], (t[1], t[2], 1 if t[1] >= thr else 0, 1))))
    sums = pairs.reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1], a[2] + b[2], a[3] + b[3]))
    rows = [(k, round(s / n, 4), round(e / n, 4), round(p / n, 4), n)
            for k, (s, e, p, n) in sums.collect()]
    return pd.DataFrame(rows, columns=["robotics_platform", "avg_score", "avg_errors",
                                       "pass_rate", "n"]).sort_values("robotics_platform")


def read_lines_rdd():
    return (sc.textFile(str(CSV_PATH), minPartitions=n_partitions)
              .filter(lambda l: not l.startswith("student_id"))
              .map(parse_line))


# ============ 4.2. Ленивые вычисления и план ===================================
t0 = time.perf_counter()
lazy_query = spark_df_stats(read_csv_df().filter(F.col("age_group") == "Средняя (11-14)"), THR)
t_build = time.perf_counter() - t0
t0 = time.perf_counter()
lazy_rows = lazy_query.collect()
t_action = time.perf_counter() - t0
print(f"Построение плана (трансформации): {t_build * 1000:.1f} мс — данные ещё не прочитаны")
print(f"Выполнение (действие collect)   : {t_action * 1000:.1f} мс\n")
print("Физический план (обратите внимание на PushedFilters и Exchange = shuffle):")
lazy_query.explain(mode="formatted")

rdd_debug = read_lines_rdd().map(lambda t: (t[0], 1)).reduceByKey(lambda a, b: a + b).toDebugString()
print("\nLineage RDD (отступы — границы стадий):")
print(rdd_debug.decode() if isinstance(rdd_debug, bytes) else rdd_debug)


# ============ 4.3. Бенчмарк =====================================================
def measure(label, fn, note):
    runs, result = [], None
    with ResourceMonitor() as mon:
        for _ in range(REPEATS):
            t0 = time.perf_counter()
            result = fn()
            runs.append(time.perf_counter() - t0)
    BENCH[label] = {"family": "Spark", "time_s": statistics.median(runs), "runs": runs,
                    "peak_rss_mb": to_mb(mon.peak_rss), "write_mb": to_mb(mon.write_bytes),
                    "tmp_files_mb": None, "cpu_s": None, "note": note}
    res_pd = result if isinstance(result, pd.DataFrame) else result.toPandas()
    print(f"{label:<34} медиана {statistics.median(runs):6.3f} с   "
          f"прогоны: {', '.join(f'{r:.3f}' for r in runs)}   "
          f"расхождение с pandas: {max_diff(res_pd.reset_index(drop=True)):.1e}")
    return res_pd


# кэшированные наборы (пересоздаются при смене числа партиций)
for name in ("sdf_cached", "rdd_cached"):
    if globals().get(name) is not None:
        globals()[name].unpersist()
sdf_cached = read_csv_df().repartition(n_partitions).cache()
rdd_cached = read_lines_rdd().repartition(n_partitions).cache()
sdf_cached.count(); rdd_cached.count()           # материализация кэша

spark_df_result = measure("Spark DataFrame (холодный)",
                          lambda: spark_df_stats(read_csv_df(), THR).toPandas(), "чтение CSV")
measure("Spark DataFrame (cache)",
        lambda: spark_df_stats(sdf_cached, THR).toPandas(), f"{n_partitions} партиций")
measure("Spark SQL (cache)",
        lambda: spark_sql_stats(sdf_cached, THR).toPandas(), f"{n_partitions} партиций")
measure("Spark RDD (холодный)",
        lambda: spark_rdd_stats(read_lines_rdd(), THR), "textFile")
measure("Spark RDD (cache)",
        lambda: spark_rdd_stats(rdd_cached, THR), f"{n_partitions} партиций")

sizes = sdf_cached.rdd.glom().map(len).collect()
print(f"\nПартиции кэшированного DataFrame: {len(sizes)}; строк в партиции: "
      f"min={min(sizes)}, max={max(sizes)}")
print(f"Партиции «холодного» чтения CSV: {read_csv_df().rdd.getNumPartitions()}")
display(spark_df_result)
````

## Ячейка 5. Сравнительный бенчмарк и визуализация

Сводим все замеры в один график и одну таблицу.

**Как читать результаты честно:**

* `mrjob` и Streaming запускают **отдельные процессы Python** на каждую стадию; в их время входят старт интерпретатора и запись промежуточных файлов;
* время Spark не включает однократный старт JVM (ячейка 1), но первый «холодный» запрос может включать прогрев;
* пиковая память для Spark — это **ядро Python + JVM**, для MapReduce — дерево процессов задания;
* объём записи — число байт, переданных в системные вызовы записи. Для короткоживущих процессов это оценка снизу;

## Ячейка 5B. Масштабирование: при каком объёме данных выигрывает Spark или MapReduce

На 50–100 тыс. строк Spark проигрывает. Причина не в «медленном» Spark: у каждой системы есть **постоянные накладные расходы**, которые на малом объёме больше самой работы. Время решения задачи можно приближённо описать моделью

$$
T(N) = a + b \cdot N,
$$

где $a$ — фиксированная часть (планирование DAG, запуск задач, обмен Python ↔ JVM, запуск процессов и `sort`), $b$ — стоимость обработки одной строки.

| | $a$ (фиксированная часть) | $b$ (стоимость строки) |
|---|---|---|
| **MapReduce** (Streaming на Python) | мала: несколько процессов и `sort` | велика: каждая строка разбирается интерпретатором Python и дважды пишется на диск |
| **Spark DataFrame** | больше: оптимизатор, планировщик, стадии, shuffle | мала: разбор CSV и агрегация идут в JVM в скомпилированном коде |

Отсюда **точка перелома** $N^* = \dfrac{a_{Spark} - a_{MR}}{b_{MR} - b_{Spark}}$. Ниже $N^*$ быстрее MapReduce, выше — Spark.

### Что делает ячейка

1. Увеличивает датасет в $k$ раз: тело CSV записывается подряд $k$ раз. Средние по платформам не меняются, а число строк растёт ровно в $k$ раз, поэтому результат легко проверить.
2. На каждом объёме измеряет:
   * **MapReduce (Streaming)**;
   * **Spark DataFrame, холодный запуск** (чтение CSV с диска);
   * **Spark с кэшем**: отдельно время построения кэша и время одного запроса;
   * **pandas** (чтение и расчёт);
   * по желанию **mrjob** (он медленный, поэтому ограничен `mrjob_max_rows`).
3. Находит точку перелома по измерениям и по модели $T = a + bN$.
4. Моделирует **итеративный сценарий**: методист двигает ползунок порога `iterative_queries` раз. MapReduce каждый раз перечитывает диск, Spark один раз строит кэш. Находится число запросов, после которого кэш окупается.
5. Оценивает, до какого объёма данные ещё помещаются в память для pandas.

### Где MapReduce остаётся уместным

На одной машине Hadoop выигрывает только на малых объёмах. В реальных кластерах его выбирают не за скорость, а за другое:

* данные во много раз больше суммарной памяти кластера;
* многочасовые пакетные задания, где важен перезапуск с сохранённого на диск результата;
* дешёвое оборудование с малым объёмом памяти;
* уже существующая инфраструктура Hadoop.

> Параметры: `scale_max_millions` — максимальный объём в миллионах строк. При 2 млн строк расчёт занимает 2–4 минуты, при 10 млн — 10–15 минут и около 3 ГБ диска.

````python
#@title Ячейка 5B. Масштабирование: где Spark обгоняет MapReduce { display-mode: "form" }
scale_max_millions = 2.0    #@param {type:"slider", min:0.5, max:10, step:0.5}
include_pandas = True       #@param {type:"boolean"}
include_mrjob = False       #@param {type:"boolean"}
mrjob_max_rows = 300000     #@param {type:"integer"}
iterative_queries = 10      #@param {type:"slider", min:2, max:30, step:1}

import matplotlib.pyplot as plt

thr_s = float(final_score_threshold)
SCALE_DIR = Path("data_scale")
shutil.rmtree(SCALE_DIR, ignore_errors=True)
SCALE_DIR.mkdir()
BASE_ROWS = len(df)
BASE_MB = to_mb(CSV_PATH.stat().st_size)
TARGETS = [50_000, 100_000, 250_000, 500_000, 1_000_000, 2_000_000,
           3_000_000, 5_000_000, 10_000_000]
limit = int(scale_max_millions * 1_000_000)
factors = sorted({max(1, round(t / BASE_ROWS)) for t in TARGETS if t <= limit} | {1})

free_gb = shutil.disk_usage(".").free / 2**30
need_gb = BASE_MB * max(factors) * 4 / 1024      # файл + копии стадий Streaming
print(f"Объёмы: {', '.join(nf(BASE_ROWS * k) for k in factors)} строк; "
      f"максимум ≈{BASE_MB * max(factors):.0f} МБ CSV. "
      f"Нужно диска ≈{need_gb:.1f} ГБ, свободно {free_gb:.0f} ГБ.")
if need_gb > 0.8 * free_gb:
    raise RuntimeError("Недостаточно места на диске: уменьшите scale_max_millions")

header, body = CSV_PATH.read_bytes().split(b"\n", 1)     # body заканчивается переводом строки


def make_scaled(k):
    path = SCALE_DIR / f"robotics_x{k}.csv"
    with open(path, "wb") as f:
        f.write(header + b"\n")
        for _ in range(k):
            f.write(body)
    return path


base_ref = (df.assign(passed=(df["final_score"] >= thr_s).astype(int))
              .groupby("robotics_platform")
              .agg(avg_score=("final_score", "mean"),
                   avg_errors=("telemetry_errors_count", "mean"),
                   pass_rate=("passed", "mean"),
                   n=("final_score", "size"))
              .round(4).reset_index())


def run_streaming(path):
    work = SCALE_DIR / "streaming_work"
    cmd = f'bash streaming_job.sh "{path}" {thr_s} {N_SPLITS} "{work}" "{sys.executable}"'
    r = run_monitored(cmd, shell=True, env=env)
    if r["returncode"] != 0:
        raise RuntimeError(r["stderr"][-300:])
    res = parse_mr_output((work / "out" / "part-00000").read_text(encoding="utf-8"))
    shutil.rmtree(work, ignore_errors=True)
    return res


def run_mrjob(path):
    tmp = SCALE_DIR / "mr_tmp"
    cmd = [sys.executable, "mr_platform_stats.py", "-r", "local", "--local-tmp-dir", str(tmp),
           "--threshold", str(thr_s), str(path)]
    r = run_monitored(cmd, env=env)
    shutil.rmtree(tmp, ignore_errors=True)
    if r["returncode"] != 0:
        err = [l for l in r["stderr"].splitlines() if l.strip()]
        raise RuntimeError(err[-1] if err else "mrjob error")
    return parse_mr_output(r["stdout"])


def run_spark_cold(path):
    return spark_df_stats(spark.read.csv(str(path), header=True, schema=SCHEMA), thr_s).toPandas()


def run_pandas(path):
    p = pd.read_csv(path, usecols=["robotics_platform", "telemetry_errors_count", "final_score"])
    return (p.assign(passed=(p["final_score"] >= thr_s).astype(int))
              .groupby("robotics_platform")
              .agg(avg_score=("final_score", "mean"),
                   avg_errors=("telemetry_errors_count", "mean"),
                   pass_rate=("passed", "mean"),
                   n=("final_score", "size"))
              .round(4).reset_index())


def timed(fn, arg, repeats):
    runs, res = [], None
    for _ in range(repeats):
        t0 = time.perf_counter()
        res = fn(arg)
        runs.append(time.perf_counter() - t0)
    return statistics.median(runs), res


SHORT = {"MapReduce: Streaming": "MR-Streaming", "MapReduce: mrjob": "mrjob",
         "Spark DataFrame (холодный)": "Spark", "Pandas (чтение + расчёт)": "pandas"}
saved_conf = {key: spark.conf.get(key) for key in
              ("spark.sql.files.maxPartitionBytes", "spark.sql.shuffle.partitions")}
spark.conf.set("spark.sql.files.maxPartitionBytes", str(64 * 2**20))
spark.conf.set("spark.sql.shuffle.partitions", str(max(2, sc.defaultParallelism)))
records = []
try:
    run_spark_cold(CSV_PATH)                      # прогрев JVM, в замеры не входит
    for k in factors:
        rows = BASE_ROWS * k
        path = make_scaled(k)
        size_mb = to_mb(path.stat().st_size)
        ref_k = base_ref.assign(n=base_ref["n"] * k)
        reps = 3 if rows <= 500_000 else 1
        engines = [("MapReduce: Streaming", run_streaming),
                   ("Spark DataFrame (холодный)", run_spark_cold)]
        if include_pandas:
            engines.append(("Pandas (чтение + расчёт)", run_pandas))
        if include_mrjob and MRJOB_OK and rows <= mrjob_max_rows:
            engines.append(("MapReduce: mrjob", run_mrjob))
        parts = [f"{nf(rows):>10} строк ({size_mb:5.0f} МБ):"]
        for name, fn in engines:
            try:
                t, res = timed(fn, path, reps)
                diff = max_diff(res, ref_k)
            except Exception as e:
                print(f"   ⚠ {name}: {type(e).__name__}: {str(e)[-200:]}")
                continue
            records.append({"engine": name, "rows": rows, "size_mb": size_mb,
                            "time_s": t, "max_diff": diff})
            parts.append(f"{SHORT[name]} {t:6.2f} с")
        cached = spark.read.csv(str(path), header=True, schema=SCHEMA).cache()
        t0 = time.perf_counter()
        cached.count()
        t_cache = time.perf_counter() - t0
        t_query, _ = timed(lambda _: spark_df_stats(cached, thr_s).toPandas(), None, 3)
        cached.unpersist()
        records.append({"engine": "Spark: запрос к кэшу", "rows": rows, "size_mb": size_mb,
                        "time_s": t_query, "max_diff": None, "cache_build_s": t_cache})
        parts.append(f"| кэш: построение {t_cache:.2f} с, запрос {t_query:.2f} с")
        print("  ".join(parts))
        path.unlink()
finally:
    for key, val in saved_conf.items():
        spark.conf.set(key, val)

SCALE_RESULTS = pd.DataFrame(records)
pivot = SCALE_RESULTS.pivot_table(index="rows", columns="engine", values="time_s")
print(f"\nМаксимальное расхождение результатов с эталоном: "
      f"{SCALE_RESULTS['max_diff'].dropna().max():.1e}")
display(pivot.round(3))


# ---------- точка перелома: по измерениям и по модели T = a + b·N -------------------
def measured_crossover(fast, slow):
    """Объём, начиная с которого `fast` быстрее `slow` (лог-интерполяция)."""
    if fast not in pivot or slow not in pivot:
        return None, "нет данных"
    ratio = (pivot[fast] / pivot[slow]).dropna()
    if ratio.empty:
        return None, "нет данных"
    if ratio.iloc[0] < 1:
        return float(ratio.index[0]), "быстрее уже на минимальном объёме"
    items = list(ratio.items())
    for (n0, r0), (n1, r1) in zip(items, items[1:]):
        if r0 >= 1 > r1:
            x0, x1, y0, y1 = np.log(n0), np.log(n1), np.log(r0), np.log(r1)
            return float(np.exp(x0 - y0 * (x1 - x0) / (y1 - y0))), "по измерениям"
    return None, f"не обогнал до {nf(ratio.index[-1])} строк"


models = {}
for eng in pivot.columns:
    s = pivot[eng].dropna()
    if len(s) >= 3:
        b, a = np.polyfit(s.index.values.astype(float), s.values, 1)
        models[eng] = (a, b)


def model_crossover(fast, slow):
    if fast not in models or slow not in models:
        return None
    (a_f, b_f), (a_s, b_s) = models[fast], models[slow]
    if b_s <= b_f:
        return None
    n_star = (a_f - a_s) / (b_s - b_f)
    return n_star if n_star > 0 else 0.0


print("\nМодель T = a + b·N (МНК по вашим замерам):")
for eng, (a, b) in models.items():
    print(f"   {eng:<28} a = {a:7.2f} с   b = {b * 1e6:7.2f} с на 1 млн строк")

MR, SP, PD = "MapReduce: Streaming", "Spark DataFrame (холодный)", "Pandas (чтение + расчёт)"
n_meas, how = measured_crossover(SP, MR)
n_model = model_crossover(SP, MR)

# ---------- итеративный сценарий: K запросов подряд ----------------------------------
iter_k = max([k for k in factors if BASE_ROWS * k <= 1_000_000] or [factors[0]])
iter_rows = BASE_ROWS * iter_k
path = make_scaled(iter_k)
spark.conf.set("spark.sql.files.maxPartitionBytes", str(64 * 2**20))
try:
    cached = spark.read.csv(str(path), header=True, schema=SCHEMA).cache()
    t0 = time.perf_counter()
    cached.count()
    t_build = time.perf_counter() - t0
    q_times = []
    for thr_i in np.linspace(40, 85, iterative_queries):
        t0 = time.perf_counter()
        spark_df_stats(cached, float(thr_i)).toPandas()
        q_times.append(time.perf_counter() - t0)
    cached.unpersist()
finally:
    spark.conf.set("spark.sql.files.maxPartitionBytes", saved_conf["spark.sql.files.maxPartitionBytes"])
    path.unlink()
    shutil.rmtree(SCALE_DIR, ignore_errors=True)

K = np.arange(1, iterative_queries + 1)
cum_spark_cache = t_build + np.cumsum(q_times)
cum_mr = pivot.loc[iter_rows, MR] * K if MR in pivot else None
cum_spark_cold = pivot.loc[iter_rows, SP] * K if SP in pivot else None
k_star = None
if cum_mr is not None:
    wins = np.where(cum_spark_cache < cum_mr)[0]
    k_star = int(K[wins[0]]) if len(wins) else None

# ---------- графики ----------------------------------------------------------------
STYLE = {MR: ("#C00000", "o", "-"), "MapReduce: mrjob": ("#E07070", "s", "-"),
         SP: ("#002060", "o", "-"), "Spark: запрос к кэшу": ("#5B7BC0", "^", "--"),
         PD: ("#8C8C8C", "D", ":")}
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))
ax = axes[0]
for eng in pivot.columns:
    s = pivot[eng].dropna()
    color, marker, ls = STYLE.get(eng, ("black", "x", "-"))
    ax.plot(s.index, s.values, marker=marker, ls=ls, color=color, label=eng)
ax.set_xscale("log"); ax.set_yscale("log")
if n_meas and how == "по измерениям":
    x_lo, x_hi = pivot.index.min() * 0.8, pivot.index.max() * 1.25
    ax.axvspan(x_lo, n_meas, color="#C00000", alpha=0.06)
    ax.axvspan(n_meas, x_hi, color="#002060", alpha=0.06)
    ax.set_xlim(x_lo, x_hi)
    ax.axvline(n_meas, color="black", ls="--", lw=1)
    right_half = np.log(n_meas) > (np.log(x_lo) + np.log(x_hi)) / 2
    ax.text(n_meas, ax.get_ylim()[0] * 1.15, f" перелом ≈ {nf(n_meas)} строк ",
            va="bottom", ha="right" if right_half else "left", fontsize=10,
            bbox=dict(facecolor="white", edgecolor="none", alpha=0.85))
ax.set_xlabel("строк в датасете (лог. шкала)")
ax.set_ylabel("время решения задачи, с (лог. шкала)")
ax.set_title("Время и объём данных")
ax.grid(True, which="both", alpha=0.3)
ax.legend(fontsize=9)

ax = axes[1]
if cum_mr is not None:
    ax.plot(K, cum_mr, "o-", color="#C00000", label="MapReduce: каждый запрос читает диск (оценка)")
if cum_spark_cold is not None:
    ax.plot(K, cum_spark_cold, "o-", color="#8CA0D0", label="Spark без кэша")
ax.plot(K, cum_spark_cache, "^-", color="#002060", label="Spark: кэш + запросы (измерено)")
if k_star and k_star > 1:
    ax.axvline(k_star, color="black", ls="--", lw=1)
ax.set_xlabel("число запросов подряд (методист двигает ползунок порога)")
ax.set_ylabel("суммарное время, с")
ax.set_title(f"Итеративный сценарий, {nf(iter_rows)} строк")
ax.grid(True, alpha=0.3)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

# ---------- выводы по фактическим числам --------------------------------------------
bytes_per_row = df.memory_usage(deep=True).sum() / len(df)
avail = psutil.virtual_memory().available
n_ram = avail / (bytes_per_row * 2.5)          # запас ×2.5 на промежуточные копии

print("Выводы по вашему эксперименту:")
if n_meas is None:
    print(f"• Spark не обогнал MapReduce ({how}). Увеличьте scale_max_millions.")
elif how == "по измерениям":
    print(f"• Spark DataFrame обгоняет MapReduce начиная примерно с {nf(n_meas)} строк "
          f"(≈{n_meas / BASE_ROWS * BASE_MB:.0f} МБ CSV).")
else:
    print(f"• Spark DataFrame {how} ({nf(n_meas)} строк).")
if n_model is not None:
    print(f"• Модель T = a + b·N даёт точку перелома ≈ {nf(n_model)} строк.")
if MR in models and SP in models:
    print(f"• Ниже точки перелома MapReduce выигрывает за счёт меньшей фиксированной части "
          f"(a: {models[MR][0]:.2f} с против {models[SP][0]:.2f} с); выше — Spark за счёт меньшей "
          f"стоимости строки ({models[SP][1] * 1e6:.2f} против {models[MR][1] * 1e6:.2f} с на 1 млн строк).")
if MR in pivot and SP in pivot:
    top = pivot.index.max()
    print(f"• На {nf(top)} строк: MapReduce {pivot.loc[top, MR]:.1f} с, Spark {pivot.loc[top, SP]:.1f} с "
          f"(в {pivot.loc[top, MR] / pivot.loc[top, SP]:.1f} раза).")
if k_star:
    when = ("уже первый запрос вместе с построением кэша быстрее MapReduce" if k_star == 1
            else f"кэш Spark окупается с {k_star}-го запроса")
    print(f"• Итеративный сценарий на {nf(iter_rows)} строк: {when} "
          f"(построение кэша {t_build:.2f} с, запрос в среднем {np.mean(q_times):.2f} с).")
elif cum_mr is not None:
    print(f"• Итеративный сценарий: за {iterative_queries} запросов кэш Spark не окупился — "
          "увеличьте iterative_queries или объём данных.")
if PD in pivot:
    pd_n, pd_how = measured_crossover(SP, PD)
    if pd_n is not None and pd_how == "по измерениям":
        print(f"• Spark обгоняет pandas примерно с {nf(pd_n)} строк.")
    else:
        print(f"• pandas на одной машине быстрее Spark на всех проверенных объёмах: "
              f"у pandas нет распределённых накладных расходов.")
print(f"• Но pandas держит все данные в памяти: ≈{bytes_per_row:.0f} байт на строку, "
      f"поэтому при {avail / 2**30:.1f} ГБ свободной RAM предел — около {nf(n_ram)} строк. "
      f"Дальше нужны Spark или MapReduce, которые обрабатывают данные по частям и умеют сбрасывать их на диск.")
print(f"• Замеры выполнены на одной машине (логических ядер: {os.cpu_count()}). В кластере фиксированная часть "
      "Spark больше, но и выигрыш на больших объёмах растёт с числом узлов.")
````

## Ячейка 6. Педагогический анализ данных (интерактивно)

Ячейка пересчитывается **автоматически** при изменении любого параметра формы:

* `final_score_threshold` — порог успешности. Ниже порога — «отсев»;
* `selected_platform`, `selected_age_group` — срез данных;
* `n_partitions` — число партиций Spark для кэшированного набора;
* `high_error_level` — сколько ошибок отладки считать «высоким уровнем»;
* `run_partition_sweep` — дополнительно замерить время запроса при 1, 2, 4 … 64 партициях.

Тяжёлая агрегация выполняется в Spark, в pandas передаются только маленькие итоговые таблицы для графиков. Это правильная привычка при работе с большими данными.

**Вопрос методиста:** *какая робототехническая платформа вызывает наибольшие трудности (ошибки датчиков и отладки) у школьников 11–14 лет?*

````python
#@title Ячейка 6. Интерактивная аналитика для методиста { run: "auto", display-mode: "form" }
final_score_threshold = 60                #@param {type:"slider", min:0, max:100, step:5}
selected_platform = "Arduino"             #@param ["Все платформы", "Lego SPIKE", "Arduino", "РОББО", "VEX", "Raspberry Pi"]
selected_age_group = "Средняя (11-14)"    #@param ["Все возрасты", "Младшая (7-10)", "Средняя (11-14)", "Старшая (15-17)"]
n_partitions = 8                          #@param {type:"slider", min:1, max:64, step:1}
high_error_level = 5                      #@param {type:"slider", min:2, max:12, step:1}
run_partition_sweep = False               #@param {type:"boolean"}

import matplotlib.pyplot as plt

THR = float(final_score_threshold)
spark.conf.set("spark.sql.shuffle.partitions", str(n_partitions))

# --- кэш под выбранное число партиций --------------------------------------------
if globals().get("_an_parts") != n_partitions:
    if globals().get("an_df") is not None:
        an_df.unpersist()
    an_df = read_csv_df().repartition(n_partitions).cache()
    an_df.count()
    _an_parts = n_partitions

age_cond = F.lit(True) if selected_age_group == "Все возрасты" else F.col("age_group") == selected_age_group
plat_cond = (F.lit(True) if selected_platform == "Все платформы"
             else F.col("robotics_platform") == selected_platform)
slice_df = an_df.filter(age_cond & plat_cond)

passed = (F.col("final_score") >= THR).cast("double")
high_err = (F.col("telemetry_errors_count") >= high_error_level).cast("double")

# --- 6.1. Ключевые показатели среза ----------------------------------------------------
t0 = time.perf_counter()
kpi = slice_df.agg(
    F.count(F.lit(1)).alias("n"),
    F.avg("final_score").alias("avg_score"),
    F.avg(passed).alias("pass_rate"),
    F.avg("telemetry_errors_count").alias("avg_err"),
    F.avg(high_err).alias("high_err_share"),
    F.expr("percentile_approx(task_completion_time_sec, 0.5)").alias("median_time"),
    F.avg("qualified_next_stage").alias("qualified_rate"),
).first()
t_query = time.perf_counter() - t0

print(f"Срез: {selected_platform} × {selected_age_group}; порог успешности {THR:g}; "
      f"партиций: {an_df.rdd.getNumPartitions()}; запрос: {t_query * 1000:.0f} мс")
if kpi["n"] == 0:
    print("В срезе нет данных — измените параметры.")
else:
    print(f"  участий: {nf(kpi['n'])}")
    print(f"  средний балл: {kpi['avg_score']:.1f}")
    print(f"  набрали ≥ {THR:g}: {kpi['pass_rate']:.1%}  →  отсев по порогу: {1 - kpi['pass_rate']:.1%}")
    print(f"  фактически прошли на следующий этап: {kpi['qualified_rate']:.1%}")
    print(f"  ошибок отладки в среднем: {kpi['avg_err']:.2f}; "
          f"≥ {high_error_level} ошибок у {kpi['high_err_share']:.1%} участников")
    print(f"  медианное время сборки и программирования: {kpi['median_time'] / 60:.1f} мин")

# --- 6.2. Данные для графиков (агрегация в Spark) ----------------------------------------
by_platform = (an_df.filter(age_cond).groupBy("robotics_platform")
               .agg(F.avg("telemetry_errors_count").alias("avg_err"),
                    F.avg(high_err).alias("high_err_share"),
                    F.avg(passed).alias("pass_rate"),
                    F.count(F.lit(1)).alias("n"))
               .orderBy(F.desc("avg_err")).toPandas())
hist = (slice_df.select((F.floor(F.col("final_score") / 5) * 5).alias("bin"))
        .groupBy("bin").count().orderBy("bin").toPandas())
STAGES_ORDER = ["Школьный", "Муниципальный", "Региональный", "Всероссийский"]
by_stage = (slice_df.groupBy("competition_stage")
            .agg(F.avg(passed).alias("pass_rate"),
                 F.avg("qualified_next_stage").alias("qualified_rate"),
                 F.count(F.lit(1)).alias("n"))
            .toPandas().set_index("competition_stage").reindex(STAGES_ORDER))
heat = (an_df.groupBy("robotics_platform", "age_group")
        .agg(F.avg("telemetry_errors_count").alias("avg_err")).toPandas()
        .pivot(index="robotics_platform", columns="age_group", values="avg_err")
        .reindex(columns=["Младшая (7-10)", "Средняя (11-14)", "Старшая (15-17)"]))

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
ax = axes[0, 0]
colors = ["#C00000" if p == selected_platform else "#002060" for p in by_platform["robotics_platform"]]
ax.bar(by_platform["robotics_platform"], by_platform["avg_err"], color=colors)
for i, r in by_platform.iterrows():
    ax.text(i, r["avg_err"], f"{r['avg_err']:.2f}\n({r['high_err_share']:.0%} ≥ {high_error_level})",
            ha="center", va="bottom", fontsize=9)
ax.set_title(f"Среднее число ошибок отладки: {selected_age_group}")
ax.set_ylabel("ошибок на участие")
ax.set_ylim(0, by_platform["avg_err"].max() * 1.3)

ax = axes[0, 1]
if len(hist):
    ax.bar(hist["bin"], hist["count"], width=4.5, align="edge",
           color=["#C00000" if b + 5 <= THR else "#002060" for b in hist["bin"]])
ax.axvline(THR, color="black", ls="--", lw=1.5)
ax.text(THR, ax.get_ylim()[1] * 0.95, f"  порог {THR:g}", va="top")
ax.set_title("Распределение итогового балла в срезе (красное — ниже порога)")
ax.set_xlabel("final_score"); ax.set_ylabel("участий")

ax = axes[1, 0]
x = np.arange(len(STAGES_ORDER))
ax.bar(x - 0.2, by_stage["pass_rate"].fillna(0), 0.4, color="#002060", label=f"балл ≥ {THR:g}")
ax.bar(x + 0.2, by_stage["qualified_rate"].fillna(0), 0.4, color="#C00000", label="прошли дальше")
ax.set_xticks(x, STAGES_ORDER)
ax.set_ylim(0, 1.05)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
ax.set_title("Успешность по этапам соревнований (срез)")
ax.legend()

ax = axes[1, 1]
im = ax.imshow(heat.values, cmap="Reds", aspect="auto")
ax.set_xticks(range(heat.shape[1]), heat.columns)
ax.set_yticks(range(heat.shape[0]), heat.index)
for i in range(heat.shape[0]):
    for j in range(heat.shape[1]):
        v = heat.values[i, j]
        ax.text(j, i, f"{v:.2f}", ha="center", va="center",
                color="white" if v > np.nanmean(heat.values) else "black")
ax.set_title("Среднее число ошибок: платформа × возраст")
fig.colorbar(im, ax=ax, shrink=0.8)
plt.tight_layout()
plt.show()

# --- 6.3. Ответ на вопрос методиста (11–14 лет) ------------------------------------
middle = (an_df.filter(F.col("age_group") == "Средняя (11-14)")
          .groupBy("robotics_platform")
          .agg(F.round(F.avg("telemetry_errors_count"), 2).alias("ошибок в среднем"),
               F.round(F.avg(high_err) * 100, 1).alias(f"% с ≥{high_error_level} ошибками"),
               F.round(F.avg("final_score"), 1).alias("средний балл"),
               F.round(F.avg(passed) * 100, 1).alias(f"% с баллом ≥{THR:g}"),
               F.count(F.lit(1)).alias("участий"))
          .orderBy(F.desc("ошибок в среднем")).toPandas())
overall_mid = an_df.filter(F.col("age_group") == "Средняя (11-14)").agg(
    F.avg("telemetry_errors_count")).first()[0]
top, easiest = middle.iloc[0], middle.iloc[-1]

print("\n=== Вывод для методиста: школьники 11–14 лет ===")
display(middle)
print(f"Наибольшие трудности вызывает {top['robotics_platform']}: {top['ошибок в среднем']:.2f} "
      f"ошибки отладки на участие при среднем по возрасту {overall_mid:.2f}; "
      f"{top[f'% с ≥{high_error_level} ошибками']:.1f}% участников допускают ≥ {high_error_level} ошибок.")
print(f"Меньше всего ошибок — на платформе {easiest['robotics_platform']} "
      f"({easiest['ошибок в среднем']:.2f}). Разница — в {top['ошибок в среднем'] / easiest['ошибок в среднем']:.1f} раза.")
print("Рекомендации:")
print(f"  1) перед работой с {top['robotics_platform']} добавить занятие по калибровке датчиков и пошаговой отладке;")
print("  2) выдавать шаблоны кода с проверкой показаний датчиков и чек-лист отладки;")
print("  3) сравнить результаты групп после изменения методики — это уже A/B-эксперимент.")
print("Осторожно: это корреляция. Платформу часто выбирает кружок, а не ученик, поэтому "
      "различия могут объясняться составом групп и опытом педагога.")

# --- 6.4. Эксперимент с партициями ------------------------------------------------------
if run_partition_sweep:
    sweep = []
    base = read_csv_df().cache(); base.count()
    for p in sorted({1, 2, 4, 8, 16, 32, 64, n_partitions}):
        spark.conf.set("spark.sql.shuffle.partitions", str(p))
        part_df = base.repartition(p).cache(); part_df.count()
        runs = []
        for _ in range(3):
            t0 = time.perf_counter()
            spark_df_stats(part_df.filter(age_cond), THR).collect()
            runs.append(time.perf_counter() - t0)
        sweep.append((p, statistics.median(runs)))
        part_df.unpersist()
    base.unpersist()
    spark.conf.set("spark.sql.shuffle.partitions", str(n_partitions))
    sw = pd.DataFrame(sweep, columns=["partitions", "time_s"])
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(sw["partitions"], sw["time_s"], marker="o", color="#002060")
    ax.scatter([n_partitions], sw.loc[sw["partitions"] == n_partitions, "time_s"],
               color="#C00000", s=120, zorder=3, label="выбрано")
    ax.set_xscale("log", base=2)
    ax.set_xlabel("число партиций"); ax.set_ylabel("медиана времени, с")
    ax.set_title(f"Время агрегирующего запроса и число партиций (логических ядер: {os.cpu_count()})")
    ax.legend(); plt.show()
    best = sw.loc[sw["time_s"].idxmin()]
    print(f"Минимальное время {best['time_s']:.3f} с при {int(best['partitions'])} партициях.")
````

### Часть A. Выявление аномальных школ

**Педагогическая гипотеза.** Если команда школы собирает и программирует робота **заметно быстрее** сверстников, но при этом допускает **заметно больше** ошибок калибровки и отладки, это может быть симптомом:

* использования чужого готового кода, который не подходит к конкретному роботу и датчикам («списывание»);
* методического перекоса: заучивание сборки без понимания программы.

**Почему нельзя сравнивать «в лоб».** Сборка на Lego SPIKE у младших и на Raspberry Pi у старших — разные задачи. Поэтому каждое участие сравнивается с **похожими** участиями через **оконные функции**:

$$
z_{time} = \frac{t - \overline{t}_{w}}{\sigma_{t,w}}, \qquad
z_{err} = \frac{e - \overline{e}_{w}}{\sigma_{e,w}}, \qquad
w = \text{окно по } (\text{robotics\_platform},\ \text{age\_group}).
$$

Дальше строится профиль школы:

| Поле | Смысл |
|---|---|
| `n_rows` | число участий |
| `mean_z_time` | средняя z-оценка времени (отрицательная — быстрее сверстников) |
| `mean_z_err` | средняя z-оценка ошибок (положительная — ошибок больше) |
| `share_fast_buggy` | доля участий, где одновременно `z_time < -1` и `z_err > 1` |
| `risk_index` | `mean_z_err − mean_z_time` |
| `risk_rank` | место школы по `risk_index` среди **всех** школ (`dense_rank` по окну без `partitionBy`) |

Школа считается подозрительной, если `n_rows ≥ MIN_ROWS`, `mean_z_time ≤ Z_TIME_MAX` и `mean_z_err ≥ Z_ERR_MIN`.

**Что нужно сделать.** Заполните строки с пометкой `TODO` в функции `find_suspicious_schools`. Каждая заглушка `None` проверяется функцией `_check`, поэтому ячейка сама подскажет, какой шаг не выполнен. Затем запустите автотест: он сравнивает найденные школы со скрытым ключом (хранятся только хэши).

> ⚖️ **Этика.** Результат — это **сигнал для методической беседы**, а не доказательство нечестности. В отчёте для администрации не указывайте персональные данные учеников (152-ФЗ) и обязательно предложите альтернативные объяснения.

````python
#@title Ячейка 7A. Поиск аномальных школ (эталонное решение) { display-mode: "form" }
MIN_ROWS = 50        #@param {type:"integer"}
Z_TIME_MAX = -0.5    #@param {type:"number"}
Z_ERR_MIN = 0.5      #@param {type:"number"}


def _check(value, step):
    if value is None:
        raise NotImplementedError(f"Не выполнен шаг TODO {step}")


def find_suspicious_schools(frame, min_rows=MIN_ROWS, z_time_max=Z_TIME_MAX, z_err_min=Z_ERR_MIN):
    # TODO 1
    w = Window.partitionBy("robotics_platform", "age_group")
    _check(w, 1)

    # TODO 2–3
    z_time_expr = ((F.col("task_completion_time_sec") - F.avg("task_completion_time_sec").over(w))
                   / F.stddev("task_completion_time_sec").over(w))
    z_err_expr = ((F.col("telemetry_errors_count") - F.avg("telemetry_errors_count").over(w))
                  / F.stddev("telemetry_errors_count").over(w))
    _check(z_time_expr, 2); _check(z_err_expr, 3)

    scored = frame.withColumn("z_time", z_time_expr).withColumn("z_err", z_err_expr)

    # TODO 4
    mean_z_time_expr = F.avg("z_time").alias("mean_z_time")
    mean_z_err_expr = F.avg("z_err").alias("mean_z_err")
    share_expr = F.avg(((F.col("z_time") < -1) & (F.col("z_err") > 1)).cast("double")).alias("share_fast_buggy")
    _check(mean_z_time_expr, "4a"); _check(mean_z_err_expr, "4b"); _check(share_expr, "4c")

    profile = (scored.groupBy("school_id")
               .agg(F.count(F.lit(1)).alias("n_rows"),
                    mean_z_time_expr, mean_z_err_expr, share_expr,
                    F.avg("final_score").alias("avg_score"))
               .withColumn("risk_index", F.col("mean_z_err") - F.col("mean_z_time")))

    # TODO 5a
    rank_w = Window.orderBy(F.desc("risk_index"))
    _check(rank_w, "5a")
    ranked = profile.withColumn("risk_rank", F.dense_rank().over(rank_w))

    # TODO 5b
    filter_cond = ((F.col("n_rows") >= min_rows)
                   & (F.col("mean_z_time") <= z_time_max)
                   & (F.col("mean_z_err") >= z_err_min))
    _check(filter_cond, "5b")

    return ranked.filter(filter_cond).orderBy("risk_rank")


base_df = an_df if globals().get("an_df") is not None else sdf
suspicious = find_suspicious_schools(base_df)
display(suspicious.toPandas().round(3))
````

````python
#@title Ячейка 7B. Автотест и карта школ { display-mode: "form" }
import matplotlib.pyplot as plt

key = json.loads(KEY_PATH.read_text(encoding="utf-8"))
expected_cols = {"school_id", "n_rows", "mean_z_time", "mean_z_err", "share_fast_buggy",
                 "avg_score", "risk_index", "risk_rank"}
missing = expected_cols - set(suspicious.columns)
assert not missing, f"В результате нет столбцов: {missing}"

found = [r["school_id"] for r in suspicious.select("school_id").collect()]
planted = set(key["anomaly_hashes"])
hits = [s for s in found if hashlib.sha256((key["salt"] + s).encode()).hexdigest() in planted]
precision = len(hits) / len(found) if found else 0.0
recall = len(hits) / key["n_anomalies"]
print(f"Найдено школ: {len(found)}; из них заложенных аномалий: {len(hits)} из {key['n_anomalies']}")
print(f"precision = {precision:.2f}, recall = {recall:.2f}")

# карта всех школ: без фильтров
all_schools = find_suspicious_schools(base_df, min_rows=0, z_time_max=float("inf"),
                                      z_err_min=float("-inf")).toPandas()
flag = all_schools["school_id"].isin(found)
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(all_schools.loc[~flag, "mean_z_time"], all_schools.loc[~flag, "mean_z_err"],
           s=all_schools.loc[~flag, "n_rows"] / 8, color="#8C8C8C", alpha=0.6, label="остальные школы")
ax.scatter(all_schools.loc[flag, "mean_z_time"], all_schools.loc[flag, "mean_z_err"],
           s=all_schools.loc[flag, "n_rows"] / 8, color="#C00000", label="подозрительные")
ax.axvline(Z_TIME_MAX, color="#C00000", ls="--", lw=1)
ax.axhline(Z_ERR_MIN, color="#C00000", ls="--", lw=1)
ax.set_xlabel("mean_z_time (левее — быстрее сверстников)")
ax.set_ylabel("mean_z_err (выше — больше ошибок)")
ax.set_title("Профили школ; размер точки — число участий")
ax.legend(); plt.show()

assert recall >= 0.8, "Найдено меньше 80% заложенных аномалий: проверьте окна и z-оценки"
assert precision >= 0.7, "Слишком много ложных срабатываний: проверьте условие отбора"
print("✅ Автотест пройден")
````

### Часть B. Предиктивная аналитика отсева

Модель логистической регрессии (Spark MLlib) предсказывает `qualified_next_stage` — пройдёт ли участник на следующий этап. Сравниваются три варианта:

1. **Базовое правило** «меньше ошибок — выше шанс»: оценкой служит `−telemetry_errors_count`. Базовая линия простая, но сильная — модель должна её обойти.
2. **Честная модель:** платформа, возраст, этап, время, ошибки и флаг «подозрительной школы» из части A.
3. **Модель с утечкой:** в признаки добавлен `final_score`. Цель вычисляется из балла, поэтому AUC будет почти идеальным. На практике такой модели нельзя пользоваться: балл становится известен только после соревнования, когда прогноз уже не нужен.

Разбиение на обучающую и тестовую выборки делается **по ученикам** (хэш `student_id`), чтобы участия одного ученика не попали в обе части.

**Для самостоятельного исследования:** сравните AUC честной модели с флагом школы и без него (уберите `school_risk_flag` из `honest`) и объясните результат.

````python
#@title Ячейка 7C. Прогноз отсева в Spark MLlib { display-mode: "form" }
from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import BinaryClassificationEvaluator
import matplotlib.pyplot as plt

base_df = globals().get("base_df")
if base_df is None:
    base_df = an_df if globals().get("an_df") is not None else sdf

try:
    flags = suspicious.select("school_id", F.lit(1).alias("school_risk_flag"))
except NameError:
    print("Часть A не выполнена — флаг школы будет нулевым.")
    flags = spark.createDataFrame([], "school_id string, school_risk_flag int")

data = (base_df.join(F.broadcast(flags), "school_id", "left")
        .fillna({"school_risk_flag": 0})
        .withColumn("label", F.col("qualified_next_stage").cast("double"))
        .withColumn("log_time", F.log1p("task_completion_time_sec"))
        .withColumn("errors", F.col("telemetry_errors_count").cast("double"))
        .withColumn("neg_errors", -F.col("errors"))
        .withColumn("score", F.col("final_score").cast("double")))
is_test = (F.abs(F.hash("student_id")) % 5) == 0
train, test = data.filter(~is_test).cache(), data.filter(is_test).cache()
print(f"train: {nf(train.count())}  test: {nf(test.count())}")

CAT = ["robotics_platform", "age_group", "competition_stage"]
evaluator = BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderROC")


def fit_and_score(num_cols):
    stages = [
        StringIndexer(inputCols=CAT, outputCols=[c + "_idx" for c in CAT], handleInvalid="keep"),
        OneHotEncoder(inputCols=[c + "_idx" for c in CAT], outputCols=[c + "_ohe" for c in CAT]),
        VectorAssembler(inputCols=[c + "_ohe" for c in CAT] + num_cols, outputCol="features"),
        LogisticRegression(featuresCol="features", labelCol="label", maxIter=50, regParam=0.01),
    ]
    model = Pipeline(stages=stages).fit(train)
    return evaluator.evaluate(model.transform(test), {evaluator.rawPredictionCol: "rawPrediction"})


honest = ["log_time", "errors", "school_risk_flag"]
results = {
    "Правило: −ошибки": evaluator.evaluate(test, {evaluator.rawPredictionCol: "neg_errors"}),
    "LogReg (честные признаки)": fit_and_score(honest),
    "LogReg + final_score (утечка!)": fit_and_score(honest + ["score"]),
}
auc = pd.Series(results, name="AUC").round(4)
display(auc.to_frame())

fig, ax = plt.subplots(figsize=(8, 3.5))
ax.barh(auc.index, auc.values, color=["#8C8C8C", "#002060", "#C00000"])
for i, v in enumerate(auc.values):
    ax.text(v, i, f" {v:.3f}", va="center")
ax.set_xlim(0.5, 1.05)
ax.set_xlabel("ROC AUC на тестовых учениках")
ax.set_title("Прогноз прохода на следующий этап")
plt.show()

gain = auc["LogReg (честные признаки)"] - auc["Правило: −ошибки"]
print(f"Честная модель лучше базового правила на {gain:+.3f} AUC.")
print(f"Модель с final_score даёт AUC {auc['LogReg + final_score (утечка!)']:.3f}: это утечка цели, "
      "а не качество модели.")
_ = (train.unpersist(), test.unpersist())
````

## Заключение

В работе одна и та же аналитическая задача решена в двух архитектурах:

* **Hadoop MapReduce** — строгий конвейер `map → shuffle/sort → reduce` с материализацией на диск. Надёжен и хорошо масштабируется для редких тяжёлых пакетных заданий, но многословен и медленен в итеративных и интерактивных сценариях;
* **Apache Spark** — ленивый DAG, кэширование в памяти и богатый API (DataFrame, SQL, оконные функции, MLlib). Позволяет строить интерактивную аналитику для методиста и предиктивные модели в одном инструменте.

Для школьного проекта важен вывод о **масштабе**: на десятках тысяч строк pandas часто быстрее обоих. Big Data-инструменты нужны, когда данные региона или платформы за годы не помещаются в память одной машины, а также когда обработку нужно распределить и защитить от отказов.
